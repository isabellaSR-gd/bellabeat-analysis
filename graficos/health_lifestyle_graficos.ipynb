```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns
import seaborn as sns
```

```python
# File path made easier 
caminho_arquivo = Path('clean_new_data_obesity.csv').resolve().parents[1]
print('caminho do diretório',caminho_arquivo)
df = pd.read_csv(f'{caminho_arquivo}/database/health-lifestyle-data/health_activity_data_treated.csv')
```

```python
sns.set_style('dark')
custom_palette=["#2e0e31", "#5D2F61", "#754B79" ,"#97709B", "#B397B6"]
sns.set_palette(custom_palette)
```

```python
df_fem = df.loc[df['Gender'] == 'Female']
```

### Gráfico 1 - Passos por faixa etária

```python
plt.figure(figsize=(7,4))
sns.scatterplot(data= df_fem,
              x= 'Daily_Steps', 
              y = 'Age')
plt.title('Passos por faixa etária')
plt.xlabel('Passos')
plt.show()
```

```python
daily_steps_age = pd.pivot_table(df_fem, values= 'Daily_Steps', columns= ('Age_Group'), 
               aggfunc={'Daily_Steps':'median'}).round(-1).astype(int)
```

```python
daily_steps_age
```

## Exercício por faixa etária

```python
plt.figure(figsize=(7,4))
sns.scatterplot(data= df_fem,
              x= 'Exercise_Hours_per_Week', 
              y = 'Age')
plt.title('Horas de exercício por faixa etária')
plt.xlabel('Horas de exercício')
plt.show()
```

```python
daily_steps_age.plot.bar()
```

```python
daily_exercise_age = pd.pivot_table(df_fem, values= 'Exercise_Hours_per_Week', columns= ('Age_Group'), 
               aggfunc={'Exercise_Hours_per_Week':'median'}).round(1)
```

```python
daily_exercise_age.plot.bar()
```
